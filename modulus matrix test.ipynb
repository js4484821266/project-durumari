```python
import numpy as np

N = 0x100
```

```python
reverse_pairs = []
for i in range(2, N):
    for j in range(i, N):
        if i * j % N == 1:
            reverse_pairs.append((i, j))
sorted(reverse_pairs, key=(lambda x: bin(x[0]).count("1") * bin(x[1]).count("1")))
```

```python
from sympy import isprime

reverse_pairs = [(a, b) for a, b in reverse_pairs if isprime(a) and isprime(b)]
reverse_pairs
```
