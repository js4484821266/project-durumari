# base, seq length, final power


```python
nn = []
for n in range(1, p):
    t = [1]
    u = n
    while u > 1:
        t.append(u)
        u = u * n % p
    nn.append((n, len(t), u))
snn = sorted(nn, key=(lambda x: (x[2] != 1, -x[1], bin(x[0]).count("1"), x[0])))
```

# LCM of length of each sequence


```python
thqo = lcm([x[1] for x in nn])
thqo
```

# Vandermonde


```python
import numpy as np

# Generate Vandermonde matrix using the bases from nn
bases = [i + 1 for i in range(thqo)]
vandermonde = np.array(
    [[pow(base, i, p) for i in range(thqo)] for base in bases], dtype=int
)
vandermonde
```

```python
# Calculate determinant of Vandermonde matrix modulo p
# For a Vandermonde matrix with bases b_1, b_2, ..., b_n:
# det = product of (b_i - b_j) for all i > j

det = 1
for i in range(len(bases)):
    for j in range(i):
        det = (det * (bases[i] - bases[j])) % p

print(f"Determinant of Vandermonde matrix mod {p}: {det}")
print(f"Inverse of the determinant: {pow(det,-1,p)}")
```

```python
#p = 257
N = p - 1  # 256


def ff_transform(f, p=257):
    """
    정변환:
        F[a] = -sum_n f[n] * a^(-n) mod p

    반환값:
        F[a-1] = a^n 기저의 계수
        a = 1, 2, ..., p-1
    """
    N = p - 1

    if len(f) != N:
        raise ValueError(f"input length must be {N}")

    F = []

    for a in range(1, p):
        a_inv = pow(a, -1, p)

        total = 0
        power = 1  # a^0

        for n in range(N):
            total += f[n] * power
            power = (power * a_inv) % p

        F.append((-total) % p)

    return F


def ff_inverse(F, p=257):
    """
    역변환:
        f[n] = sum_a F[a] * a^n mod p

    즉 a^n들이 직접 합성 기저가 된다.
    """
    N = p - 1

    if len(F) != N:
        raise ValueError(f"spectrum length must be {N}")

    f = []

    for n in range(N):
        total = 0

        for a in range(1, p):
            total += F[a - 1] * pow(a, n, p)

        f.append(total % p)

    return f


# --------------------------------------------------
# 테스트 신호
# --------------------------------------------------

f = [n % 5 for n in range(N)]

F = ff_transform(f, p)
recovered = ff_inverse(F, p)


# 0이 아닌 스펙트럼만 출력
print("Nonzero spectrum:")

for a, value in enumerate(F, start=1):
    if value != 0:
        print(f"a={a:3d}, coefficient={value}")


print()
print("Original :", f[:])
print("Recovered:", recovered[:])
print("Exact reconstruction:", f == recovered)
```
