```python
t = n = 125
ttt = []
while t != 1 and t != 0 and len(ttt) < 0x100:
    ttt.append(t)
    t = t * n % 0x100
ttt.append(1)
```

```python
len(ttt),len(sorted(list(set(ttt))))
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(12, 4))
plt.plot(ttt, color="tab:blue", linewidth=0.5)
plt.title(f"{n}^n mod 256")
plt.xlabel("n")
plt.ylabel(f"{n}^n")
plt.grid(alpha=0.2)
plt.show()
```

```python
plt.figure(figsize=(12, 4))
plt.plot(sorted(ttt), color="tab:blue", linewidth=0.5)
plt.title(f"{n}^n mod 256, sorted")
plt.xlabel("n")
plt.ylabel(f"{n}^n")
plt.grid(alpha=0.2)
plt.show()
```

```python
ttt = [[]]
for _ in range(0x100):
    ttt.append([])
    t = n = _ + 1
    while t != 1 and t != 0 and len(ttt[-1]) < 0x100:
        ttt[-1].append(t)
        t = t * n % 0x100
```

```python
lel=list(enumerate([len(a)for a in ttt]))
lel
```

```python
sorted(lel,key=lambda x:x[1],reverse=True)
```
